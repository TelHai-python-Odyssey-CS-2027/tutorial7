````python
# Import libraries
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_openml
import pandas as pd
````

# Step 1: Select a Dataset

Choose a dataset from `sklearn` / OpenML (Iris, wine, …).


### דוגמאות
**Iris** — `fetch_openml(name="iris", version=1, as_frame=True, parser="pandas")`  
**Wine** — `fetch_openml(name="wine", version=1, as_frame=True, parser="pandas")`


````python
data = fetch_openml(name="iris", version=1, as_frame=True, parser="pandas")
````

# Step 2: Inspect the Data

- `data.DESCR` — תיאור
- `df = data.frame` — הטבלה
- `df.sample(5)` — דגימה
- `df.describe()` / `df.dtypes` — סיכום וסוגים


````python
print(data.DESCR)
````

````python
df = data.frame  # DataFrame: הטבלה עצמה
df.sample(5)     # חמש שורות אקראיות
````

````python
df.describe()
````

# Step 3: Select features

קודם תדפיסו את **כל** העמודות:

```python
print(df.columns)
features = list(df.columns)
```

אחר כך בחרו כמה שמות לרשימה `selected_features` (עדיף עמודות מספריות):

```python
selected_features = [features[0], features[1], features[2], features[3]]
# או בשמות מפורשים:
# selected_features = ["sepal length (cm)", "petal length (cm)", ...]
```


````python
# 1) כל העמודות
print(df.columns)
features = list(df.columns)
print("Available features:", features)

# 2) בחירה של כמה features
selected_features = []  # למשל: [features[0], features[1], features[2], features[3]]
print("Selected features:", selected_features)
````

# Step 4: Histograms

לכל feature נבחר — היסטוגרמה. שימו לב ל־`bins`.


````python
fig, axs = plt.subplots(1, len(selected_features), figsize=(20, 3))
if len(selected_features) == 1:
    axs = [axs]

for ax, f in zip(axs, selected_features):
    ax.hist(df[f], bins=5, color="skyblue", edgecolor="black")
    ax.set_xlabel(f)
    ax.set_title(f)

plt.tight_layout()
plt.show()
````

# Step 5: One feature vs many others

בחרו **ציר Y אחד** (`reference_feature`) וציירו אותו מול **כל שאר** ה־features הנבחרים (לא רק זוג אחד).


````python
reference_feature = selected_features[0]  # או שם מחרוזת
y = df[reference_feature]

other_features = [f for f in selected_features if f != reference_feature]

fig, axs = plt.subplots(1, len(other_features), figsize=(4 * len(other_features), 3))
if len(other_features) == 1:
    axs = [axs]

for ax, f in zip(axs, other_features):
    ax.scatter(df[f], y, alpha=0.6)
    ax.set_xlabel(f)
    ax.set_ylabel(reference_feature)
    ax.set_title(f"{reference_feature} vs {f}")

plt.tight_layout()
plt.show()
````

# Step 6: Save one correlation plot

בחרו זוג מעניין ושמרו לקובץ — זה מה שמעלים ב־PS7 יחד עם `analysis.txt`.


````python
reference_feature = selected_features[0]
comparison_feature = selected_features[1]  # שנו לזוג שבחרתם

plt.figure(figsize=(8, 6))
plt.scatter(df[comparison_feature], df[reference_feature], alpha=0.6)
plt.xlabel(comparison_feature)
plt.ylabel(reference_feature)
plt.title(f"{reference_feature} vs {comparison_feature}")

plt.savefig("correlation_plot.png", dpi=150, bbox_inches="tight")
plt.show()
````
