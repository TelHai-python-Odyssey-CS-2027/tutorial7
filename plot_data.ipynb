Then choose a few interesting features (about 4–5) and put their **names** in a Python list called `selected_features`.

# by name

Tip: for scatter/correlation later, prefer **numeric** columns.


````python
# 1) See every column name
print(df.columns)
features = list(df.columns)
print("Available features:", features)

# 2) Pick a few names into selected_features
selected_features = []  # e.g. [features[0], features[1], features[2], features[3]]
print("Selected features:", selected_features)
````

# Step 4: Make Histogram Plots

For each of the selected features, create a histogram to visualize the distribution of the data. You can use `plt.hist()` or `plt.bar()` to create the plots.

Notice: The `.hist(bins=5)` function creates the histogram with 5 bins, but you can adjust this based on the dataset.



````python
fig, axs  = plt.subplots(1, len(selected_features), figsize = (20,3))

for ax, f in zip(axs, selected_features):
    ax.hist(df[f], bins=5, color='skyblue', edgecolor='black')
    ax.set_xlabel(f)
````

# Step 5: One feature vs many others

Choose **one** feature as the reference (this will be the **y-axis**).

Then plot that same y against **each of the other selected features** (many x-axes) — not just one pair.

- Use a row of subplots: one scatter per other feature.
- Look for linear / non-linear trends, clusters, and outliers.
- Skip plotting the reference against itself.


````python
# One y-axis, many x-axes
reference_feature = selected_features[0]  # or a name string, e.g. selected_features[1]
y = df[reference_feature]

other_features = [f for f in selected_features if f != reference_feature]

fig, axs = plt.subplots(1, len(other_features), figsize=(4 * len(other_features), 3))
if len(other_features) == 1:
    axs = [axs]

for ax, f in zip(axs, other_features):
    ax.scatter(df[f], y, alpha=0.6)
    ax.set_xlabel(f)
    ax.set_ylabel(reference_feature)
    ax.set_title(f"{reference_feature} vs {f}")

plt.tight_layout()
plt.show()
````

# Step 6: Choose one correlation plot and save it

From the many scatters above, pick **one** pair that looks interesting (strong / surprising / non-linear).

- Save it with `plt.savefig("correlation_plot.png")`.
- In a short note (for the homework: `analysis.txt`), describe what you see.


````python
reference_feature = selected_features[0]
comparison_feature = selected_features[1]  # change to the pair you liked

plt.figure(figsize=(8, 6))
plt.scatter(df[comparison_feature], df[reference_feature], alpha=0.6)
plt.xlabel(comparison_feature)
plt.ylabel(reference_feature)
plt.title(f"{reference_feature} vs {comparison_feature}")

plt.savefig("correlation_plot.png", dpi=150, bbox_inches="tight")
plt.show()
````
