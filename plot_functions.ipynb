# Plot functions — גרפים של פונקציות

נצייר פונקציות מוכרות ממתמטיקה (גבולות) כדי לראות התנהגות ליד נקודות «בעייתיות».

## הגדרות קצרות (חדש)

| כלי | מה זה |
| --- | --- |
| `np.linspace(a, b, n)` | יוצר רשימת/`array` של `n` מספרים בערך שווה־מרווח מ־`a` עד `b` |
| `math.sin(x)` | סינוס של `x` (רדיאנים) |
| `plt.plot(xs, ys, marker="o")` | קו המחבר נקודות; `marker` מסמן כל נקודה |
| `plt.show()` | מציג את הגרף |

```python
import numpy as np   # np.linspace — רשימת מספרים בטווח
import math          # math.sin — סינוס
import matplotlib.pyplot as plt  # ציור גרפים
```

## הבעיה 1: \(\frac{\sin x}{x}\)

ב־\(x=0\) הביטוי לא מוגדר (חלוקה באפס), אבל הגבול קיים.

**מה נעשה:** ניצור ערכי \(x\) עם `np.linspace`, נדלג על 0, נחשב `math.sin(x)/x`, ונצייר עם `plt.plot(..., marker="o")`.


```python
xs_with_zero = np.linspace(-10, 10, 100)
xs = []
for x in xs_with_zero:
    if x != 0:
        xs.append(x)

ys = []
for x in xs:
    ys.append(math.sin(x) / x)

plt.plot(xs, ys, marker="o")
plt.xlabel("x")
plt.ylabel("sin(x) / x")
plt.title("sin(x) / x")
plt.grid(True, alpha=0.3)
plt.show()
```

## הבעיה 2: \(x\sin\frac{1}{x}\)

גם כאן ב־\(x=0\) הביטוי בעייתי (\(\sin(1/x)\) מתנדנד מהר מאוד).

**שימו לב:** ליד 0 צריך יותר נקודות (`linspace` עם `n` גדול) וטווח קטן, אחרת לא רואים את התנודות.


```python
xs_with_zero = np.linspace(-0.1, 0.1, 1000)
xs = []
for x in xs_with_zero:
    if x != 0:
        xs.append(x)

ys = []
for x in xs:
    ys.append(x * math.sin(1 / x))

plt.plot(xs, ys, marker="o", markersize=2)
plt.xlabel("x")
plt.ylabel("x * sin(1/x)")
plt.title("x * sin(1/x)")
plt.grid(True, alpha=0.3)
plt.show()
```
