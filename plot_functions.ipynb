# Plot functions

We will plot familiar math functions (limits) to see their behavior near "problematic" points.

## Quick definitions (new)

| Tool | Meaning |
| --- | --- |
| `np.linspace(a, b, n)` | Creates a list/`array` of `n` roughly equally spaced numbers from `a` to `b` |
| `math.sin(x)` | Sine of `x` (radians) |
| `plt.plot(xs, ys, marker="o")` | Line connecting points; `marker` marks each point |
| `plt.show()` | Displays the plot |

```python
import numpy as np   # np.linspace — numbers in a range
import math          # math.sin — sine
import matplotlib.pyplot as plt  # plotting
```

## Problem 1: \(\frac{\sin x}{x}\)

At \(x=0\) the expression is undefined (division by zero), but the limit exists.

**What we will do:** create \(x\) values with `np.linspace`, skip 0, compute `math.sin(x)/x`, and plot with `plt.plot(..., marker="o")`.


```python
xs_with_zero = np.linspace(-10, 10, 100)
xs = []
for x in xs_with_zero:
    if x != 0:
        xs.append(x)

ys = []
for x in xs:
    ys.append(math.sin(x) / x)

plt.plot(xs, ys, marker="o")
plt.xlabel("x")
plt.ylabel("sin(x) / x")
plt.title("sin(x) / x")
plt.grid(True, alpha=0.3)
plt.show()
```

## Problem 2: \(x\sin\frac{1}{x}\)

Here too the expression is problematic at \(x=0\) (\(\sin(1/x)\) oscillates very quickly).

**Note:** near 0 you need more points (`linspace` with a large `n`) and a small range, otherwise you will not see the oscillations.


```python
xs_with_zero = np.linspace(-0.1, 0.1, 1000)
xs = []
for x in xs_with_zero:
    if x != 0:
        xs.append(x)

ys = []
for x in xs:
    ys.append(x * math.sin(1 / x))

plt.plot(xs, ys, marker="o", markersize=2)
plt.xlabel("x")
plt.ylabel("x * sin(1/x)")
plt.title("x * sin(1/x)")
plt.grid(True, alpha=0.3)
plt.show()
```
