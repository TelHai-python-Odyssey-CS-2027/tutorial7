# Plot data starter — zip, histogram, subplots, scatter

## Quick definitions (new)

| Tool | Meaning |
| --- | --- |
| `zip(xs, ys)` | Iterates over two lists **in parallel**: each round a pair `(x, y)` |
| `plt.hist(data, bins=k)` | Histogram for a continuous value; `bins` = number of bins/bars |
| `plt.title(...)` | Title for a **standalone** plot (not a subplot) |
| `ax.set_title(...)` | Title for an **axis inside** `subplots` |
| `plt.subplots(1, n)` | One row with `n` side-by-side plots → `fig, axs` |
| `ax.scatter(x, y)` | Points — to check the relationship between two variables |

```python
import random
from matplotlib import pyplot as plt
```

# Zip

`zip` pairs two (or more) iterables by position:


```python
xs = ["A", "B", "C", "D"]
ys = [1, 10, 100, 1000]

for x, y in zip(xs, ys):
    print(x, y)
```

# Histogram

A continuous value (random number between 0 and 1) → how many times it fell into each "bin" (`bins`).


```python
def generate_random_numbers(n):
    data = []
    for _ in range(n):
        data.append(random.random())
    return data


data = generate_random_numbers(100)
plt.hist(data, bins=10, edgecolor="black")
plt.title("100 random numbers")  # title for a standalone plot
plt.show()
```

# Subplots

When you have several plots together — use `ax.set_title` (not `plt.title` on each one).


```python
fig, axs = plt.subplots(1, 3, figsize=(20, 5))

data = generate_random_numbers(100)
axs[0].hist(data, bins=10, edgecolor="black")
axs[0].set_title("100 random numbers")

data = generate_random_numbers(1000)
axs[1].hist(data, bins=10, edgecolor="black")
axs[1].set_title("1000 random numbers")

data = generate_random_numbers(10000)
axs[2].hist(data, bins=10, edgecolor="black")
axs[2].set_title("10000 random numbers")

plt.show()
```

```python
n_items = [10**1, 10**2, 10**3, 10**4, 10**5]

fig, axs = plt.subplots(1, len(n_items), figsize=(20, 3))

for ax, n in zip(axs, n_items):
    data = generate_random_numbers(n)
    ax.hist(data, bins=10, edgecolor="black")
    ax.set_title(str(n) + " random numbers")

plt.show()
```

# Correlation (scatter)

Scatter shows the relationship between \(x\) and \(y\): random vs correlated.


```python
x = list(range(10))

y_random = []
for i in x:
    y_random.append(random.random())

y_corr = []
for i in x:
    y_corr.append(-i + random.random())

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].scatter(x, y_random)
axs[0].set_title("Random")
axs[0].set_xlabel("x", fontsize=16)
axs[0].set_ylabel("y", fontsize=16)

axs[1].scatter(x, y_corr)
axs[1].set_title("Correlated data")
axs[1].set_xlabel("x", fontsize=16)
axs[1].set_ylabel("y", fontsize=16)
plt.show()
```
